```python
import numpy as np
import statsmodels as sm
import matplotlib.pyplot as plt
```

```python
def generate_data(n: int):
    return np.random.uniform(size=n)

# Example
np.sort(generate_data(100))
```

```python
def pipeline(n, ax_hist, ax):
    data = generate_data(n)
    
    ax_hist.hist(data, bins=25, density=True, color='lightskyblue', edgecolor='black', alpha=0.7, label='Выборка')
    ax_hist.axhline(y=1.0, color='red', linestyle='--', lw=2, label='Теоретическая плотность')
    
    ax_hist.set_title(f'Гистограмма плотности (n = {n})')
    ax_hist.set_xlabel('x')
    ax_hist.set_ylabel('f(x)')
    ax_hist.set_ylim(0, 1.8)
    ax_hist.grid(True, linestyle=':', alpha=0.6)
    ax_hist.legend()
    
    
    x = np.sort(data)
    y = np.arange(1, n + 1) / n
    ax.step(x, y, where='post', label='Выборочная ФР', color='royalblue', lw=2)
    ax.plot([0, 1], [0, 1], '--', color='red', label='Теоретическая ФР')
    
    ax.set_title(f'Выборочная ФР (n = {n})')
    ax.set_xlabel('x')
    ax.set_ylabel('F_n(x)')
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend()
    ax.set_xlim(-0.05, 1.05)
    ax.set_ylim(-0.05, 1.05)
```

```python
sample_sizes = [10, 10**2, 10**3, 10**4]

fig, axes = plt.subplots(len(sample_sizes), 2, figsize=(12, 4 * len(sample_sizes)))

for i, n in enumerate(sample_sizes):
    pipeline(n, axes[i, 0], axes[i, 1])

plt.tight_layout()
plt.show()
```
